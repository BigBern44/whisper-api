```python
import asyncio
import aiohttp
import time

async def send_request(session, file_path, i):
    url = "http://localhost:8000/transcribe"
    with open(file_path, "rb") as f:
        data = aiohttp.FormData()
        data.add_field("file", f, filename="audio.mp3", content_type="audio/mpeg")
        async with session.post(url, data=data) as resp:
            result = await resp.json()
            print(f"[{i}] {result}")
            return result

file_path = "audio.mp3"
n_requests = 100

start = time.time()

async with aiohttp.ClientSession() as session:
    tasks = [send_request(session, file_path, i) for i in range(n_requests)]
    results = await asyncio.gather(*tasks)

elapsed = time.time() - start
print(f"\n{n_requests} requêtes en {elapsed:.2f}s ({n_requests/elapsed:.1f} req/s)")

```
